```python
import numpy as np # массивы
```

```python
MeV = 1 # Энергетическая константа
```

```python
class GeneratorSource: # Источник фотонов

    def __init__(self, Y : int):
    
        self.Y = Y

    def GeneratePhotonDirections(self, E): # изтропный генератор направлений от события с энергией E

        LambdaPoisson = int(self.Y*E) # Считаем, сколько фотонов должно образоваться
        N_photons = np.random.poisson(LambdaPoisson) # Случайным образом определяем число фотонов для генерации
        N_photons_per_set = 100 # 100 фотонов за раз

        # Определяем, сколько раз генератор сможет выдать пачку фотонов
        if ((N_photons % N_photons_per_set) == 0):
            N_sets = N_photons // N_photons_per_set
        else:
            N_sets = N_photons // N_photons_per_set + 1

        for i in range(N_sets): # Цикл генерации

            N_gen = np.min((N_photons_per_set, N_photons-i*N_photons_per_set)) # Если получилось число фотонов, не кратное 100, то в последний раз нужно выдать меньше

            # Изотропно -> равномерно по косинусу тета и равномерно по фи
            cos_theta = np.random.uniform(-1, 1, N_gen)
            sin_theta = np.sqrt(1-cos_theta**2)
            phi = np.random.uniform(0, 2*np.pi, N_gen)
    
            # Получаем единичные вектора направлений
            directions = np.array([sin_theta*np.cos(phi), sin_theta*np.sin(phi), cos_theta]).T
            
            yield directions
```

```python
MySource = GeneratorSource(10000/MeV) # Объявляем источник
```

```python
DirectionGenerator = MySource.GeneratePhotonDirections(1*MeV)
```

```python
# Проверка

for direc in DirectionGenerator:
    print(direc)
```
